```python
import pandas as pd
from google.colab import files

decades = [
    ("1986-11-13T12:00:00Z", "1996-12-31T12:00:00Z"),
    ("1997-01-01T12:00:00Z", "2006-12-31T12:00:00Z"),
    ("2007-01-01T12:00:00Z", "2016-12-31T12:00:00Z"),
    ("2017-01-01T12:00:00Z", "2026-10-29T12:00:00Z"),
]

chunks = []
for start, end in decades:
    print(f"Downloading {start[:4]} to {end[:4]}...")
    url = (
        f"https://coastwatch.pfeg.noaa.gov/erddap/griddap/ncdcOisst21Agg_LonPM180.csv?"
        f"sst%5B({start}):1:({end})%5D%5B(0.0):1:(0.0)%5D%5B(26.125):1:(26.125)%5D%5B(-80.125):1:(-80.125)%5D,"
        f"anom%5B({start}):1:({end})%5D%5B(0.0):1:(0.0)%5D%5B(26.125):1:(26.125)%5D%5B(-80.125):1:(-80.125)%5D,"
        f"err%5B({start}):1:({end})%5D%5B(0.0):1:(0.0)%5D%5B(26.125):1:(26.125)%5D%5B(-80.125):1:(-80.125)%5D"
    )
    chunk = pd.read_csv(url, skiprows=[1], parse_dates=["time"], index_col="time")
    chunks.append(chunk)

# Combine all 4 decades, drop any NaNs, and save/download the CSV
df = pd.concat(chunks).dropna()
print(f"\nDone! Total clean rows: {len(df)}")

df.to_csv("hollywood_reefs_sst.csv")
files.download("hollywood_reefs_sst.csv")
df.head()
```

```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import numpy as np

url = "https://coastwatch.pfeg.noaa.gov/erddap/griddap/ncdcOisst21Agg_LonPM180"
ds = xr.open_dataset(url)

target_dates = [
    "1986-11-15T12:00:00",
    "1996-11-15T12:00:00",
    "2006-11-15T12:00:00",
    "2016-11-15T12:00:00",
    "2026-10-29T12:00:00",
]

regional_sst = (
    ds["sst"]
    .sel(
        time=target_dates,
        method="nearest",
    )
    .sel(
        zlev=0.0,
        latitude=slice(24.5, 27.5),
        longitude=slice(-81.5, -78.5),
    )
    .load()
)

#cartopy map
vmin = float(np.floor(regional_sst.min()))
vmax = float(np.ceil(regional_sst.max()))

fig, axes = plt.subplots(
    1,
    len(target_dates),
    figsize=(20, 4.8),
    subplot_kw={"projection": ccrs.PlateCarree()},
    constrained_layout=True,
)

for i, ax in enumerate(axes):
    slice_2d = regional_sst.isel(time=i)
    date_label = str(slice_2d.time.values)[:10]

    # Plot SST field
    im = slice_2d.plot(
        ax=ax,
        transform=ccrs.PlateCarree(),
        cmap="turbo",
        vmin=vmin,
        vmax=vmax,
        add_colorbar=False,
    )

#coast and borders
    ax.add_feature(cfeature.LAND, facecolor="#e8e6e1", edgecolor="black", linewidth=0.6, zorder=3)
    ax.add_feature(cfeature.COASTLINE, linewidth=0.8, zorder=4)
    ax.set_extent([-81.5, -78.5, 24.5, 27.5], crs=ccrs.PlateCarree())

    # Pin your exact Hollywood Reefs pixel (26.125°N, -80.125°W)
    ax.plot(
        -80.125,
        26.125,
        marker="*",
        color="white",
        markeredgecolor="black",
        markersize=12,
        transform=ccrs.PlateCarree(),
        zorder=5,
        label="Hollywood Reefs",
    )

#grid lines
    gl = ax.gridlines(draw_labels=True, linewidth=0.4, color="gray", alpha=0.5, linestyle="--")
    gl.top_labels = False
    gl.right_labels = False
    if i > 0:
        gl.left_labels = False

    ax.set_title(date_label, fontsize=12, fontweight="bold")

# Shared colorbar across all 5 maps
cbar = fig.colorbar(im, ax=axes, orientation="vertical", fraction=0.015, pad=0.02)
cbar.set_label("Sea Surface Temperature (°C)", fontsize=11, fontweight="bold")

fig.suptitle(
    "Hollywood Reefs & Southeast Florida: 40-Year Sea Surface Temperature Progression",
    fontsize=15,
    fontweight="bold",
)
plt.show()
```
